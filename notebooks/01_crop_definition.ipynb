# Crop Object

This notebook introduces the `Crop` object, the basic building block used to describe each cultivated species in the IWR workflow.

## What the `Crop` object represents

The `Crop` object is a small data container that stores the agronomic parameters needed by the model for one crop type. It combines:

- the crop identity, such as `maize` or `wheat`
- water and growth parameters used by the model
- an optional raster map that shows where that crop is present in the study area

## Main fields

The object is defined as a dataclass with the following attributes:

- `name`: crop name
- `root_max_depth`: maximum rooting depth
- `kc_initial`: initial crop coefficient
- `kc_max`: maximum crop coefficient
- `kc_sen`: crop coefficient during senescence
- `p`: depletion fraction used in the water balance
- `crop_raster`: optional raster with crop fraction values

## How it is created

In the codebase, a crop can be created in two main ways:

- `Crop.from_name(...)` to load one of the predefined crops from the internal library
- `Crop.custom(...)` to start from a predefined crop and override selected parameters

## Why it matters

Later in the workflow, the `Crop` object is used together with soil and phenology information to compute irrigation water requirements and crop-specific outputs.

In the next cells, we will inspect the crop parameters step by step and build a first example crop instance.

## Exploring the predefined crop library

The `Crop` class includes a built-in library of predefined crop definitions. In the next example, we will:

- display the list of available crop names
- show a simple visual plot of those names
- print the predefined parameters for `maize`

This is a good first check before building a crop instance for a real analysis.

```python
%matplotlib inline
from iwr.crops import CROP_LIBRARY, Crop
from IPython.display import display
import matplotlib.pyplot as plt

# Show the available predefined crop names in the console.
Crop.print_crops()

# Print the predefined parameters for maize.
Crop.print_crop_parameters("maize")

# Plot the crop names as a simple visual reference for beginners.
crop_names = sorted(CROP_LIBRARY)
```

## Creating a crop from its predefined name

A simple way to build a `Crop` object is to start from one of the predefined names in the internal library.

In the next example, we will create the `maize` crop with `Crop.from_name("maize")` and print the resulting object.

This is useful because it shows the full set of parameters stored in the crop object before the model uses it.

```python
from iwr.crops import Crop

crop = Crop.from_name("maize")
print(crop)
```

## Creating a custom crop based on maize

You can also create a new crop by starting from an existing predefined crop and changing only the parameters you need.

In the next example, we create `maize_b` from `maize`, but we change:

- `p` from `0.55` to `0.6`
- `kc_max` from `1.2` to `1.3`

This is useful when you want to test a crop variant without modifying the original predefined crop.

```python
maize_b = Crop.custom("maize", name="maize_b", p=0.6, kc_max=1.3)
print(maize_b)
```
